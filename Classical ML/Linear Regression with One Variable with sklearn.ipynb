```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
x = np.array([1.0, 1.5, 2.5, 3.0, 4.0, 5.0, 6.0, 7.5, 8.0, 9.0])
y = np.array([52.0, 56.0, 61.0, 65.0, 72.0, 78.0, 84.0, 91.0, 93.0, 98.0])
X = x.reshape(-1,1)
```

```python
model =  LinearRegression()
model.fit(X, y)
```

```python
w = model.coef_[0]
b = model.intercept_

print(f"The weight = {w}")
print(f"The bias = {b}")
```

```python
predict = np.array([[6.5]])
Score = model.predict(predict)

print(f"The expected score for 6.5 hours of Study = {Score.item():.2f}")
```

```python
plt.Figure()
plt.scatter(x, y)

x_line = np.linspace(min(x), max(x))
y_line = w * x_line + b

plt.plot(x_line, y_line, color='red')
```
