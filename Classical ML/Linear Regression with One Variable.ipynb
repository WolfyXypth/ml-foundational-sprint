```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([1.0, 1.5, 2.5, 3.0, 4.0, 5.0, 6.0, 7.5, 8.0, 9.0])
y_train = np.array([52.0, 56.0, 61.0, 65.0, 72.0, 78.0, 84.0, 91.0, 93.0, 98.0])
```

```python
def compute_cost(x, y, w, b):
    
    m = x.shape[0]

    total_cost = 0
    for i in range(m):
        f_wb = w * x[i] + b
        total_cost += (f_wb - y[i]) ** 2

    cost =  total_cost / (2 * m)
    return cost
```

```python
def compute_gradient(x, y, w, b):
    m = x.shape[0]

    total_dJ_dw = 0
    total_dJ_db = 0

    for i in range(m):
        f_wb = w * x[i] + b
        total_dJ_dw += (f_wb - y[i]) * x[i]
        total_dJ_db += (f_wb - y[i])

    dJ_dw = (1 / m) * total_dJ_dw
    dJ_db = (1 / m) * total_dJ_db
    return dJ_dw, dJ_db
```

```python
def gradient_descent (x, y, w, b, alpha, NumIter, cost, gradient):
    J_history =  []
    for i in range(NumIter):
        cost_i = cost(x, y, w, b)
        J_history.append(cost_i)
        
        dJ_dw, dJ_db = gradient(x, y, w, b)
        w = w - alpha * dJ_dw
        b = b - alpha * dJ_db

    print(f"cost = {cost_i}")
    print(f"dJ_dw, dJ_db = {dJ_dw, dJ_db}")
    return w, b, J_history
```

```python
alpha = 10e-3
w = 0
b = 0
NumIter = 1000
w, b, J_history = gradient_descent(x_train, y_train, w, b, alpha, NumIter, compute_cost, compute_gradient)
print(f"w, b = {w, b}")
print(f"The expected score of a person with 6.5 hrs studied is {w * 6.5 + b}")
```

```python
plt.figure()
plt.plot(range(len(J_history)), J_history, color='purple', linewidth=2)

plt.xlabel('Iteration Number')
plt.ylabel('Cost $J(w, b)$')
plt.title('Gradient Descent Convergence (Learning Curve)')

plt.show()
```
